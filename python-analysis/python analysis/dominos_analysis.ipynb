```python
# Domino's Sales Analytics Notebook
# ==================================
# Tools: Pandas, NumPy, Matplotlib
# Purpose: Data cleaning, EDA, KPI analysis, and business insights

import os
import warnings
warnings.filterwarnings("ignore")

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

plt.style.use("default")


# ============================================================
# 1. LOAD DATA
# ============================================================

# Update this path if needed
df = pd.read_csv('D:\Data Science Projects\dominos analysis\dataset\Dominos_Sales_Data.csv')

print("Dataset loaded successfully.")
print("Shape:", df.shape)
display(df.head())


# ============================================================
# 2. CLEAN COLUMN NAMES
# ============================================================

df.columns = (
    df.columns.astype(str)
    .str.strip()
    .str.lower()
    .str.replace(" ", "_", regex=False)
    .str.replace("-", "_", regex=False)
)

print("Cleaned columns:")
print(df.columns.tolist())


# ============================================================
# 3. BASIC INSPECTION
# ============================================================

print("\nDATA INFO")
print(df.info())

print("\nDESCRIBE NUMERIC COLUMNS")
display(df.describe())

print("\nMISSING VALUES")
display(df.isna().sum().sort_values(ascending=False))

print("\nDUPLICATE ROWS:", df.duplicated().sum())


# ============================================================
# 4. DETECT IMPORTANT COLUMNS
# ============================================================

def find_col(possible_names):
    for name in possible_names:
        if name in df.columns:
            return name
    return None

COL_ORDER_ID = find_col(["order_id", "orderid"])
COL_ORDER_DATE = find_col(["order_date", "date", "orderdate"])
COL_ORDER_TIME = find_col(["order_time", "time", "ordertime"])
COL_CUSTOMER_ID = find_col(["customer_id", "customerid"])
COL_STORE_ID = find_col(["store_id", "storeid"])
COL_CITY = find_col(["city"])
COL_PIZZA_NAME = find_col(["pizza_name", "product", "item_name"])
COL_CATEGORY = find_col(["category"])
COL_SIZE = find_col(["size"])
COL_QUANTITY = find_col(["quantity", "qty"])
COL_UNIT_PRICE = find_col(["unit_price", "price", "unitprice"])
COL_DISCOUNT = find_col(["discount"])
COL_GST = find_col(["gst", "tax"])
COL_TOTAL_AMOUNT = find_col(["total_amount", "amount", "revenue", "sales"])
COL_ORDER_TYPE = find_col(["order_type", "channel"])
COL_PAYMENT_METHOD = find_col(["payment_method", "payment"])
COL_DISCOUNT_CATEGORY = find_col(["discount_category"])

print("\nDetected columns:")
for k, v in {
    "order_id": COL_ORDER_ID,
    "order_date": COL_ORDER_DATE,
    "order_time": COL_ORDER_TIME,
    "customer_id": COL_CUSTOMER_ID,
    "store_id": COL_STORE_ID,
    "city": COL_CITY,
    "pizza_name": COL_PIZZA_NAME,
    "category": COL_CATEGORY,
    "size": COL_SIZE,
    "quantity": COL_QUANTITY,
    "unit_price": COL_UNIT_PRICE,
    "discount": COL_DISCOUNT,
    "gst": COL_GST,
    "total_amount": COL_TOTAL_AMOUNT,
    "order_type": COL_ORDER_TYPE,
    "payment_method": COL_PAYMENT_METHOD,
    "discount_category": COL_DISCOUNT_CATEGORY,
}.items():
    print(f"{k}: {v}")


# ============================================================
# 5. BASIC CLEANING
# ============================================================

# Remove duplicate rows
before = len(df)
df = df.drop_duplicates().copy()
print(f"\nDropped {before - len(df)} duplicate rows.")

# Convert date column
if COL_ORDER_DATE is None:
    raise ValueError("order_date column not found. Please rename/map it correctly.")

df[COL_ORDER_DATE] = pd.to_datetime(df[COL_ORDER_DATE], errors="coerce")

# Convert time column if present
if COL_ORDER_TIME is not None:
    df[COL_ORDER_TIME] = pd.to_datetime(df[COL_ORDER_TIME].astype(str), errors="coerce").dt.time

# Convert numeric columns
for col in [COL_ORDER_ID, COL_CUSTOMER_ID, COL_STORE_ID, COL_QUANTITY, COL_UNIT_PRICE, COL_DISCOUNT, COL_GST, COL_TOTAL_AMOUNT]:
    if col is not None and col in df.columns:
        df[col] = pd.to_numeric(df[col], errors="coerce")

# Standardize text columns
text_cols = [COL_CITY, COL_PIZZA_NAME, COL_CATEGORY, COL_SIZE, COL_ORDER_TYPE, COL_PAYMENT_METHOD, COL_DISCOUNT_CATEGORY]
for col in text_cols:
    if col is not None and col in df.columns:
        df[col] = df[col].astype(str).str.strip()
        df[col] = df[col].replace({"nan": np.nan, "None": np.nan, "": np.nan})

# Drop rows missing critical fields
critical_cols = [COL_ORDER_DATE, COL_TOTAL_AMOUNT]
if COL_ORDER_ID is not None:
    critical_cols.append(COL_ORDER_ID)

df = df.dropna(subset=critical_cols).copy()

print("\nShape after cleaning:", df.shape)


# ============================================================
# 6. FEATURE ENGINEERING
# ============================================================

df["year"] = df[COL_ORDER_DATE].dt.year
df["month"] = df[COL_ORDER_DATE].dt.month
df["month_name"] = df[COL_ORDER_DATE].dt.strftime("%B")
df["day_name"] = df[COL_ORDER_DATE].dt.day_name()
df["day_number"] = df[COL_ORDER_DATE].dt.dayofweek + 1

if COL_ORDER_TIME is not None:
    df["hour"] = pd.to_datetime(df[COL_ORDER_TIME].astype(str), errors="coerce").dt.hour
else:
    df["hour"] = np.nan

df["weekend_flag"] = np.where(df["day_name"].isin(["Saturday", "Sunday"]), "Weekend", "Weekday")

def hour_bucket(hour):
    if pd.isna(hour):
        return np.nan
    hour = int(hour)
    if 6 <= hour <= 11:
        return "Morning"
    elif 12 <= hour <= 16:
        return "Afternoon"
    elif 17 <= hour <= 21:
        return "Peak Evening"
    else:
        return "Late Night"

df["hour_bucket"] = df["hour"].apply(hour_bucket)

print("\nFeature engineering completed.")
display(df.head())


# ============================================================
# 7. KPI CALCULATIONS
# ============================================================

total_revenue = df[COL_TOTAL_AMOUNT].sum()
total_orders = df[COL_ORDER_ID].nunique() if COL_ORDER_ID is not None else len(df)
aov = total_revenue / total_orders if total_orders else 0
total_pizzas_sold = df[COL_QUANTITY].sum() if COL_QUANTITY is not None else np.nan
total_customers = df[COL_CUSTOMER_ID].nunique() if COL_CUSTOMER_ID is not None else np.nan

print("\n================= KPI SUMMARY =================")
print(f"Total Revenue       : {total_revenue:,.2f}")
print(f"Total Orders        : {total_orders:,.0f}")
print(f"Average Order Value : {aov:,.2f}")
print(f"Total Pizzas Sold   : {total_pizzas_sold:,.0f}" if pd.notna(total_pizzas_sold) else "Total Pizzas Sold   : N/A")
print(f"Total Customers     : {total_customers:,.0f}" if pd.notna(total_customers) else "Total Customers     : N/A")
print("==============================================\n")


# ============================================================
# 8. HELPER FUNCTION FOR PLOTS
# ============================================================

def plot_bar(series, title, xlabel, ylabel="Revenue", figsize=(10, 5), color="steelblue", rotate=0):
    plt.figure(figsize=figsize)
    series.plot(kind="bar", color=color)
    plt.title(title, fontsize=14, fontweight="bold")
    plt.xlabel(xlabel)
    plt.ylabel(ylabel)
    plt.xticks(rotation=rotate)
    plt.tight_layout()
    plt.show()

def plot_line(series, title, xlabel, ylabel="Revenue", figsize=(10, 5), color="steelblue", marker="o"):
    plt.figure(figsize=figsize)
    series.plot(kind="line", marker=marker, color=color)
    plt.title(title, fontsize=14, fontweight="bold")
    plt.xlabel(xlabel)
    plt.ylabel(ylabel)
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()


# ============================================================
# 9. MONTHLY REVENUE ANALYSIS
# ============================================================

month_order = ["January", "February", "March", "April", "May", "June",
               "July", "August", "September", "October", "November", "December"]

monthly_revenue = (
    df.groupby("month_name")[COL_TOTAL_AMOUNT]
      .sum()
      .reindex(month_order)
)

print("\nMonthly Revenue:")
display(monthly_revenue)

plot_line(monthly_revenue, "Monthly Revenue Trend", "Month", "Revenue", figsize=(12, 5), color="navy")


# ============================================================
# 10. REVENUE BY CITY
# ============================================================

if COL_CITY is not None:
    city_revenue = df.groupby(COL_CITY)[COL_TOTAL_AMOUNT].sum().sort_values(ascending=False)
    print("\nRevenue by City:")
    display(city_revenue)
    plot_bar(city_revenue.head(10), "Top Cities by Revenue", "City", "Revenue", figsize=(12, 6), color="darkred", rotate=45)


# ============================================================
# 11. TOP 5 PIZZAS BY REVENUE
# ============================================================

if COL_PIZZA_NAME is not None:
    top_pizzas = df.groupby(COL_PIZZA_NAME)[COL_TOTAL_AMOUNT].sum().sort_values(ascending=False).head(5)
    print("\nTop 5 Pizzas by Revenue:")
    display(top_pizzas)
    plot_bar(top_pizzas, "Top 5 Pizzas by Revenue", "Pizza Name", "Revenue", figsize=(10, 5), color="teal", rotate=30)


# ============================================================
# 12. TOP 5 PIZZAS BY QUANTITY
# ============================================================

if COL_PIZZA_NAME is not None and COL_QUANTITY is not None:
    top_pizzas_qty = df.groupby(COL_PIZZA_NAME)[COL_QUANTITY].sum().sort_values(ascending=False).head(5)
    print("\nTop 5 Pizzas by Quantity:")
    display(top_pizzas_qty)
    plot_bar(top_pizzas_qty, "Top 5 Pizzas by Quantity", "Pizza Name", "Quantity", figsize=(10, 5), color="purple", rotate=30)


# ============================================================
# 13. REVENUE BY CATEGORY
# ============================================================

if COL_CATEGORY is not None:
    category_revenue = df.groupby(COL_CATEGORY)[COL_TOTAL_AMOUNT].sum().sort_values(ascending=False)
    print("\nRevenue by Category:")
    display(category_revenue)
    plot_bar(category_revenue, "Revenue by Category", "Category", "Revenue", figsize=(8, 5), color="orange", rotate=0)


# ============================================================
# 14. REVENUE BY SIZE
# ============================================================

if COL_SIZE is not None:
    size_revenue = df.groupby(COL_SIZE)[COL_TOTAL_AMOUNT].sum().sort_values(ascending=False)
    print("\nRevenue by Size:")
    display(size_revenue)
    plot_bar(size_revenue, "Revenue by Pizza Size", "Size", "Revenue", figsize=(8, 5), color="forestgreen", rotate=0)


# ============================================================
# 15. REVENUE BY DAY OF WEEK
# ============================================================

day_order = ["Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday", "Sunday"]
day_revenue = df.groupby("day_name")[COL_TOTAL_AMOUNT].sum().reindex(day_order)

print("\nRevenue by Day of Week:")
display(day_revenue)
plot_bar(day_revenue, "Revenue by Day of Week", "Day", "Revenue", figsize=(10, 5), color="royalblue", rotate=20)


# ============================================================
# 16. REVENUE BY HOUR
# ============================================================

hour_revenue = df.dropna(subset=["hour"]).groupby("hour")[COL_TOTAL_AMOUNT].sum().sort_index()

print("\nRevenue by Hour:")
display(hour_revenue)
plot_line(hour_revenue, "Revenue by Hour", "Hour", "Revenue", figsize=(10, 5), color="crimson")


# ============================================================
# 17. ORDER TYPE ANALYSIS
# ============================================================

if COL_ORDER_TYPE is not None:
    order_type_revenue = df.groupby(COL_ORDER_TYPE)[COL_TOTAL_AMOUNT].sum().sort_values(ascending=False)
    print("\nRevenue by Order Type:")
    display(order_type_revenue)
    plot_bar(order_type_revenue, "Revenue by Order Type", "Order Type", "Revenue", figsize=(9, 5), color="slateblue", rotate=0)


# ============================================================
# 18. PAYMENT METHOD ANALYSIS
# ============================================================

if COL_PAYMENT_METHOD is not None:
    payment_revenue = df.groupby(COL_PAYMENT_METHOD)[COL_TOTAL_AMOUNT].sum().sort_values(ascending=False)
    print("\nRevenue by Payment Method:")
    display(payment_revenue)
    plot_bar(payment_revenue, "Revenue by Payment Method", "Payment Method", "Revenue", figsize=(9, 5), color="brown", rotate=20)


# ============================================================
# 19. DISCOUNT ANALYSIS
# ============================================================

if COL_DISCOUNT_CATEGORY is not None:
    discount_revenue = df.groupby(COL_DISCOUNT_CATEGORY)[COL_TOTAL_AMOUNT].sum().sort_values(ascending=False)
    discount_orders = df.groupby(COL_DISCOUNT_CATEGORY)[COL_ORDER_ID].nunique().sort_values(ascending=False) if COL_ORDER_ID is not None else None

    print("\nRevenue by Discount Category:")
    display(discount_revenue)
    plot_bar(discount_revenue, "Revenue by Discount Category", "Discount Category", "Revenue", figsize=(9, 5), color="darkorange", rotate=20)

    if discount_orders is not None:
        print("\nOrders by Discount Category:")
        display(discount_orders)





# ============================================================
# 20. WEEKEND VS WEEKDAY ANALYSIS
# ============================================================

weekend_sales = df.groupby("weekend_flag")[COL_TOTAL_AMOUNT].sum()
print("\nWeekend vs Weekday Revenue:")
weekend_sales = df.groupby("weekend_flag")[COL_TOTAL_AMOUNT].sum().sort_values(ascending=False)
display(weekend_sales)
plot_bar(weekend_sales, "Weekend vs Weekday Revenue", "Segment", "Revenue", figsize=(8, 5), color="magenta", rotate=0)


# ============================================================
# 21. PEAK HOUR ANALYSIS
# ============================================================

peak_hour_sales = df.groupby("hour_bucket")[COL_TOTAL_AMOUNT].sum()

print("\nRevenue by Hour Bucket:")
display(peak_hour_sales)
plot_bar(peak_hour_sales, "Revenue by Hour Bucket", "Hour Bucket", "Revenue", figsize=(8, 5), color="goldenrod", rotate=20)


# ============================================================
# 22. BUSINESS INSIGHTS
# ============================================================

print("\n================= BUSINESS INSIGHTS =================")
print(f"1. Total revenue generated by Domino’s is {total_revenue:,.2f}.")
print(f"2. Total orders placed are {total_orders:,.0f}, with an average order value of {aov:,.2f}.")
if pd.notna(total_pizzas_sold):
    print(f"3. Total pizzas sold are {total_pizzas_sold:,.0f}.")
if pd.notna(total_customers):
    print(f"4. Total unique customers are {total_customers:,.0f}.")

if COL_CITY is not None:
    top_city = city_revenue.index[0]
    print(f"5. Top revenue city is {top_city}.")

if COL_PIZZA_NAME is not None:
    top_pizza = top_pizzas.index[0]
    print(f"6. Top revenue pizza is {top_pizza}.")

top_month = monthly_revenue.idxmax()
print(f"7. Highest revenue month is {top_month}.")

top_day = day_revenue.idxmax()
print(f"8. Highest revenue day is {top_day}.")

if len(hour_revenue) > 0:
    top_hour = hour_revenue.idxmax()
    print(f"9. Peak ordering hour is {top_hour}:00.")

print("====================================================\n")


# ============================================================
# 2. EXPORT CLEANED DATA
# ============================================================

output_path = "Dominos_Sales_Cleaned.csv"
df.to_csv(output_path, index=False)
print(f"Cleaned dataset exported to: {output_path}")
```
